- Jenkins is an open-source automation tool used mainly for CI/CD in software development
- Jenkins works on master slave architecture
- Jenkins helps developers automate repetitive tasks like:
    - Build code
    - Running tests
    - Deploying Applications

- **Job:** It's a basic unit of work in jenkins, A job is something Jenkins executes
    - Types:
        - Freestyle:
            - Simple UI based configuration
        - Pipeline Job: 
            - Defined using code
            - Supports complex workflows
        - Multibranch Pipeline:
            - Automatically runs jobs for different git branches
        - Multiconfiguration Project:
            - Also called Matrix jobs
            - Run the same job in multiple environments/configuration
- **Build:** A build is a single execution of a job.
- **Pipeline:** A series of steps or stages inside a job
    - Types:
        - Declarative Pipeline (simpler)
        - Scripted Pipeline (more flexible)
- **Jenkinsfile**: It is where we define our pipeline code
- Controller: 
    - It's the main branch of jenkin, which manages jobs, schedules build, assign work to agent, and store configuration
- Agent:
    - It's a machine that actually run jobs
    - Used due to distributed work load
- Executor:
    - It's a slot inside a agent
    - 1 executer = 1 job at a time
- Workspace:
    - A workspacce is the directory where jenkins download code, run builds, and run the scripts
- Artifacts:
    - These are output files from a build
    - They can be stored and used later
- Plugins:
    - They can extend jenkins functionality


- Jenkin Pipeline code uses **Groovy Syntax** (For scripted pipeline)
- A sample systax is
```bash
----------------------------------------------
pipeline {
    agent any

    stages {
        stage('Hello') {
            steps {
                echo 'Hello Folder'
            }
        }
        stage('Create Folder') {
            steps {
                sh 'mkdir -p devops'
            }
        }
    }
}
----------------------------------------------
```
